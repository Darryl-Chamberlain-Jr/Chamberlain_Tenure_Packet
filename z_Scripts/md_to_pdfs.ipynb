```python
import os
from subprocess import call

script_dir = os.getcwd()
base_dir = os.path.dirname(script_dir)
```

```python
brief_state_path = os.path.join(base_dir, 'Application_Statement', 'Brief_Statement')
sum_teach_state_path = os.path.join(base_dir, 'Teaching_Effectiveness', 'summary_teaching_effectiveness')
sum_course_eval_path = os.path.join(base_dir, 'Teaching_Effectiveness', 'Optional_Artifacts', 'student_course_evaluation_narrative')
sum_scholar_path = os.path.join(base_dir, 'Scholarly_Activity', 'summary_scholarly_activity')
sum_service_path = os.path.join(base_dir, 'Service', 'summary_service')

list_of_paths = [brief_state_path, sum_teach_state_path, sum_course_eval_path, sum_scholar_path, sum_service_path]
pdf_path = os.path.join(base_dir, 'PDFs')
```

```python
for path in list_of_paths:
    call_list = ['pandoc',
        f'{path}.md', 
        '-o',
        f'{os.path.join(pdf_path, os.path.splitext(os.path.basename(path))[0])}.pdf',
        '-V',
        'geometry:margin=1in'
    ]
    call(call_list)
```
